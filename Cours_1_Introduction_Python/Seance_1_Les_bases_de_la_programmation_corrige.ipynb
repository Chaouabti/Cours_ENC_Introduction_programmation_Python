## **3. Tuples**

Un tuple ressemble à une liste, mais il est **immuable** : on ne peut plus le modifier après sa création. Il s'écrit entre parenthèses `( )`.

Il convient bien à une fourchette de datation, qui ne doit pas changer par erreur.

**Création d'un tuple :**

````python
datation = (1200, 1210)
print(datation)
print(datation[0])
````

**Immuabilité :** toute tentative de modification provoque une erreur. Exécutez la cellule pour la voir.

````python
datation[0] = 1190
````

**Déballage (*unpacking*) :** on assigne chaque élément à une variable distincte.

````python
debut, fin = datation
print(f"Manuscrit daté entre {debut} et {fin}")
````

## **4. Dictionnaires**

Un dictionnaire associe des **clés uniques** à des **valeurs**. Il s'écrit entre accolades `{ }`, avec `clé: valeur` pour chaque paire. C'est la forme naturelle d'une notice de catalogue.

**Création d'un dictionnaire :**

````python
manuscrit = {
    "cote": "MS 24",
    "bibliotheque": "Aberdeen, University Library",
    "date": 1200,
    "numerise": True
}
print(manuscrit)
````

**Accéder aux valeurs :**

````python
print(manuscrit["cote"])
````

**Ajouter une nouvelle paire clé-valeur :** on assigne une valeur à une clé qui n'existe pas encore.

````python
manuscrit["langue"] = "latin"
print(manuscrit)
````

Pour ajouter plusieurs paires d'un coup, on peut utiliser la méthode `update()` :

````python
manuscrit.update({"origine": "Angleterre", "support": "parchemin"})
print(manuscrit)
````

**Méthodes utiles :**
- `keys()` renvoie les clés du dictionnaire.

````python
print(manuscrit.keys())
````

- `values()` renvoie les valeurs.

````python
print(manuscrit.values())
````

- `items()` renvoie les paires clé-valeur, sous forme de tuples.

````python
print(manuscrit.items())
````

## **5. Sets (ensembles)**

Un set est une collection **non ordonnée** d'éléments **uniques**. Il s'écrit entre accolades `{ }` ou avec la fonction `set()`.

**Création d'un set :**

````python
creatures_set = {"lion", "panthère", "licorne", "dragon"}
print(creatures_set)
````

**Remarque :** pour créer un set vide, utilisez `set()` et non `{}`, qui crée un dictionnaire vide.

````python
set_vide = set()
print(set_vide)
````

**Les doublons disparaissent automatiquement.** Pratique pour savoir quelles créatures sont mentionnées dans un texte, sans compter les répétitions :

````python
mentions = ["lion", "dragon", "lion", "licorne", "dragon"]
print(set(mentions))
````

##### **Méthodes courantes**
- `add(element)` ajoute un élément.
- `remove(element)` supprime un élément, avec une erreur s'il n'existe pas.
- `discard(element)` supprime un élément, sans erreur s'il n'existe pas.

````python
creatures_set.add("phénix")
creatures_set.remove("licorne")
creatures_set.discard("sirène")   # "sirène" n'est pas dans le set : pas d'erreur
print(creatures_set)
````

##### **Opérations ensemblistes**

Les sets permettent de comparer deux collections. Voici deux extraits simplifiés de tables des chapitres :

````python
bestiaire_1 = {"lion", "panthère", "licorne", "dragon", "phénix"}
bestiaire_2 = {"lion", "phénix", "sirène", "serpent", "licorne"}
````

- **Union** : toutes les créatures présentes dans l'un ou l'autre.

````python
print(bestiaire_1.union(bestiaire_2))
````

- **Intersection** : les créatures communes aux deux.

````python
print(bestiaire_1.intersection(bestiaire_2))
````

- **Différence** : les créatures du premier absentes du second.

````python
print(bestiaire_1.difference(bestiaire_2))
````

- **Différence symétrique** : les créatures présentes dans un seul des deux.

````python
print(bestiaire_1.symmetric_difference(bestiaire_2))
````

## **7. Opérateurs et conditions**

#### **Opérateurs de comparaison**

`==`, `!=`, `>`, `<`, `>=`, `<=` : ils renvoient `True` ou `False`.

````python
print(manuscrit["date"] == 1200)
print(manuscrit["date"] < 1200)
print(manuscrit["langue"] != "français")
````

#### **Opérateurs logiques**

`and` (toutes les conditions vraies), `or` (au moins une vraie), `not` (inverse).

````python
print(manuscrit["date"] <= 1200 and manuscrit["numerise"])
print(manuscrit["langue"] == "latin" or manuscrit["date"] > 1300)
print(not manuscrit["numerise"])
````

#### **Structure `if`, `elif`, `else`**

Classer un manuscrit par siècle. L'ordre des conditions compte : la première condition vraie l'emporte.

````python
date = 1150

if date <= 1100:
    print("XIe siècle ou avant")
elif date <= 1200:
    print("XIIe siècle")
elif date <= 1300:
    print("XIIIe siècle")
else:
    print("Après 1300")
````

Si l'on inverse l'ordre, le résultat devient faux :

````python
if date <= 1300:
    print("XIIIe siècle")
elif date <= 1200:
    print("XIIe siècle")
````

## **8. Boucles bornées : `for`**

Parcourir une séquence (liste, chaîne, etc.) élément par élément.

**Exemple avec une liste :**

````python
for creature in creatures:
    print(creature)
````

**Utilisation de `range()` :** `range(début, fin, pas)`, la fin étant exclue.

````python
for f in range(1, 6):
    print(f"f. {f}")
````

````python
for f in range(1, 10, 2):
    print(f"f. {f}")
````

**Instructions `break` et `continue` :**
- `break` sort immédiatement de la boucle.
- `continue` passe directement à l'itération suivante.

````python
for creature in creatures:
    if creature == "licorne":
        print("Licorne trouvée, arrêt de la recherche.")
        break
````

````python
for mot in titre.split(" "):
    if len(mot) <= 2:
        continue   # on ignore les mots courts
    print(mot)
````

## **9. Boucles non bornées : `while`**

La boucle `while` répète des instructions **tant qu'une condition est vraie**. Elle convient quand on ne sait pas à l'avance combien de tours il faudra.

````python
mots_titre = titre.split(" ")
i = 0
while mots_titre[i] != "colombe.":
    i += 1
print(f"« colombe » est à l'index {i}")
````

**Attention aux boucles infinies :** si la condition reste toujours vraie, la boucle ne s'arrête jamais. Il faut que la condition finisse par devenir fausse, ou prévoir un `break`.

```python
while True:
    print("Ceci est une boucle infinie !")   # ne pas exécuter !
```

Les instructions `break` et `continue` fonctionnent aussi avec `while`.
