En séance 1, nous avons classé un manuscrit par siècle avec `if/elif/else`. Plaçons ce code dans une fonction : on pourra alors l'appliquer à n'importe quelle date.

### **1.3 Renvoyer ou afficher : `return` et `print()`**

- `print()` **affiche** une valeur à l'écran : la valeur n'est pas conservée.
- `return` **renvoie** une valeur au programme : on peut la ranger dans une variable et la réutiliser.

Voici la même fonction, écrite avec `print()` au lieu de `return` :

````python
def siecle_affiche(date):
    """Afficher le siècle d'une date."""
    if date <= 1100:
        print("XIe siècle ou avant")
    elif date <= 1200:
        print("XIIe siècle")
    elif date <= 1300:
        print("XIIIe siècle")
    else:
        print("Après 1300")


valeur = siecle_affiche(1200)   # affiche XIIe siècle...
print(valeur)                   # ... mais la variable contient None
````

Une fonction sans `return` renvoie quand même quelque chose : la valeur spéciale `None`, qui signifie « rien ». Comparons les **types** des deux résultats :

````python
print(type(siecle(1200)))           # une chaîne de caractères
print(type(siecle_affiche(1200)))   # NoneType : le type de None
````

Conséquence : le résultat de `siecle_affiche()` ne peut pas être réutilisé, par exemple dans une phrase.

````python
phrase = f"Le manuscrit date du {siecle_affiche(1200)}."
print(phrase)

phrase = f"Le manuscrit date du {siecle(1200)}."
print(phrase)
````

### **1.4 Paramètres, arguments et valeurs par défaut**

Les **paramètres** sont les noms déclarés dans `def`. Les **arguments** sont les valeurs transmises lors de l'appel. Un paramètre peut avoir une **valeur par défaut**, utilisée si l'appel ne fournit pas d'argument pour lui.

````python
def creer_notice(cote, bibliotheque, date, langue="latin"):
    """Renvoyer la notice d'un manuscrit sous forme de dictionnaire."""
    return {"cote": cote, "bibliotheque": bibliotheque,
            "date": date, "langue": langue}


# Arguments positionnels : associés aux paramètres dans l'ordre
print(creer_notice("MS 24", "Aberdeen, University Library", 1200))

# Arguments nommés : l'ordre n'a plus d'importance
print(creer_notice(date=1210, cote="MS. Ashmole 1511",
                   bibliotheque="Oxford, Bodleian Library"))

# Remplacer la valeur par défaut
print(creer_notice("MS 24", "Aberdeen, University Library", 1200,
                   langue="anglo-normand"))
````

### **1.6 Nombre variable d'arguments : `*args` et `**kwargs`**

- `*` regroupe les arguments positionnels supplémentaires dans un **tuple** ;
- `**` regroupe les arguments nommés supplémentaires dans un **dictionnaire**.

````python
def decrire_creatures(cote, *noms):
    """Renvoyer une phrase sur les créatures d'un manuscrit."""
    return f"{cote} : {len(noms)} créatures {noms}"


print(decrire_creatures("MS 24", "dragon", "licorne", "phénix"))
````

````python
print(decrire_creatures("MS 24", creatures))    # la liste entière compte pour 1
print(decrire_creatures("MS 24", *creatures))   # les 5 créatures sont transmises une à une
````

## **2. Documenter son code**

Le code dit **ce que** fait le programme ; le commentaire explique **pourquoi**. Une **docstring** décrit une fonction ou un module : Python la conserve et `help()` l'affiche.

Voici la docstring complète de `siecle()`, selon les conventions de la PEP 257 : une première ligne de résumé, une ligne vide, puis les détails.

````python
def siecle(date):
    """Renvoyer le siècle d'une date sous forme de texte.

    Un siècle va de l'an 1 à l'an 100 : 1200 appartient au XIIe siècle.

    Paramètre :
        date (int) : année de copie du manuscrit, par exemple 1200.

    Renvoie :
        str : le siècle, par exemple "XIIe siècle".
    """
    if date <= 1100:
        resultat = "XIe siècle ou avant"
    elif date <= 1200:
        # 1200 est inclus : c'est la dernière année du XIIe siècle
        resultat = "XIIe siècle"
    elif date <= 1300:
        resultat = "XIIIe siècle"
    else:
        resultat = "Après 1300"
    return resultat


help(siecle)
````

`help()` fonctionne aussi avec les fonctions de Python.

Dans VS Code, la docstring s'affiche au survol du nom de la fonction. Le raccourci `Shift + Option + A` (Mac) ou `Shift + Alt + A` (Windows, Linux) entoure la sélection de triples guillemets.

````python
help(len)
````

#### **Deuxième méthode : la commande magique `%%writefile`**

Jupyter propose des **commandes magiques**, qui commencent par `%` ou `%%` et ne font pas partie du langage Python.

Placée en première ligne d'une cellule, `%%writefile nom_du_fichier.py` **n'exécute pas** le code de la cellule : elle l'**enregistre dans un fichier**.

- Si le fichier existe déjà, il est **écrasé** : la cellule ci-dessous remplace donc le `bestiaire.py` que vous venez de créer, par un contenu identique.
- `%%writefile -a nom_du_fichier.py` **ajoute** le contenu à la fin du fichier au lieu de l'écraser.
- Le code n'étant pas exécuté, les fonctions ne sont disponibles qu'après un `import`.

````python
%%writefile bestiaire.py
"""Fonctions et données pour le corpus de bestiaires médiévaux."""

CREATURES = ["dragon", "licorne", "lion", "panthère", "phénix"]


def siecle(date):
    """Renvoyer le siècle d'une date sous forme de texte."""
    if date <= 1100:
        resultat = "XIe siècle ou avant"
    elif date <= 1200:
        resultat = "XIIe siècle"
    elif date <= 1300:
        resultat = "XIIIe siècle"
    else:
        resultat = "Après 1300"
    return resultat


def cote_complete(manuscrit):
    """Renvoyer la bibliothèque et la cote d'un manuscrit."""
    return f"{manuscrit['bibliotheque']}, {manuscrit['cote']}"
````
