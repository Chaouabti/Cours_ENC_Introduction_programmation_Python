#### **Deuxième méthode : la commande magique `%%writefile`**

Jupyter propose des **commandes magiques**, qui commencent par `%` ou `%%` et ne font pas partie du langage Python.

Placée en première ligne d'une cellule, `%%writefile nom_du_fichier.py` **n'exécute pas** le code de la cellule : elle l'**enregistre dans un fichier**.

- Si le fichier existe déjà, il est **écrasé** : la cellule ci-dessous remplace donc le `bestiaire.py` que vous venez de créer, par un contenu identique.
- `%%writefile -a nom_du_fichier.py` **ajoute** le contenu à la fin du fichier au lieu de l'écraser.
- Le code n'étant pas exécuté, les fonctions ne sont disponibles qu'après un `import`.

````python
%%writefile bestiaire.py
"""Fonctions et données pour le corpus de bestiaires médiévaux."""

CREATURES = ["dragon", "licorne", "lion", "panthère", "phénix"]


def siecle(date):
    """Renvoyer le siècle d'une date sous forme de texte."""
    if date <= 1100:
        resultat = "XIe siècle ou avant"
    elif date <= 1200:
        resultat = "XIIe siècle"
    elif date <= 1300:
        resultat = "XIIIe siècle"
    else:
        resultat = "Après 1300"
    return resultat


def cote_complete(manuscrit):
    """Renvoyer la bibliothèque et la cote d'un manuscrit."""
    return f"{manuscrit['bibliotheque']}, {manuscrit['cote']}"
````

````python
from bestiaire import siecle, CREATURES

print(siecle(1150))
print(len(CREATURES))
````

### **1.6 Organiser ses modules : les packages** *(facultatif)*

Quand un module grandit, on peut le découper en plusieurs fichiers rangés dans un dossier : c'est un **package**. Créons un package `bestiaires` (avec un *s*, pour le distinguer de notre module `bestiaire.py`), qui contiendra un module `dates.py`.

`%%writefile` ne crée pas les dossiers : on crée d'abord le dossier avec `pathlib`.

````python
Path("bestiaires").mkdir(exist_ok=True)   # exist_ok=True : pas d'erreur si le dossier existe déjà
````

````python
%%writefile bestiaires/dates.py
"""Fonctions de datation."""


def siecle(date):
    """Renvoyer le siècle d'une date sous forme de texte."""
    if date <= 1100:
        resultat = "XIe siècle ou avant"
    elif date <= 1200:
        resultat = "XIIe siècle"
    elif date <= 1300:
        resultat = "XIIIe siècle"
    else:
        resultat = "Après 1300"
    return resultat
````

Le fichier `__init__.py` signale que le dossier est un package. Il est le plus souvent vide ; ici, il importe `siecle` pour raccourcir les importations. Le point de `.dates` signifie « le module `dates` de ce même package ».

## **3. Classes et objets**

Jusqu'ici, un manuscrit est un **dictionnaire**, et les fonctions qui le traitent, comme `cote_complete(manuscrit)`, sont rangées à part. Avec la programmation orientée objet, nous allons créer nos propres **objets**, qui réunissent les données (les **attributs**) et les actions (les **méthodes**).

### **3.1 Vous utilisez déjà des objets**

En Python, toutes les données sont des objets. `type()` indique la **classe** d'un objet : `titre` est une instance de la classe `str`, et `split()` est une méthode de cette classe.

````python
print(type(titre))       # <class 'str'>
print(type(creatures))   # <class 'list'>
print(type(manuscrit))   # <class 'dict'>
````

### **3.2 Une première classe**

Le mot-clé `class` définit une classe. Son nom s'écrit en **PascalCase** : chaque mot commence par une majuscule.

Le bloc d'une classe ne peut pas être vide. `pass` signifie « ne rien faire » : il permet de créer une classe sans contenu, que l'on complétera plus tard.

Pour créer un objet, on appelle la classe avec des parenthèses, comme une fonction : c'est l'**instanciation**.

````python
class Manuscrit:
    pass


manuscrit = Manuscrit()
print(manuscrit)
print(type(manuscrit))
````

`print()` affiche la classe de l'objet et son adresse en mémoire, qui change à chaque exécution. `__main__` désigne le programme en cours, ici le notebook.

**Attention** : la variable `manuscrit` désigne maintenant un objet `Manuscrit`, et non plus le dictionnaire de la séance 1. Python respecte la casse : `Manuscrit` (la classe) et `manuscrit` (l'objet) sont deux noms différents.

Sans `pass`, Python refuse la classe. Décommentez pour voir l'erreur, puis recommentez.

## **4. Attributs et méthodes**

### **4.1 La méthode `__init__`, le paramètre `self` et les attributs d'instance**

- `__init__` est appelée **automatiquement** à la création de l'objet : elle initialise ses attributs.
- `self` désigne **l'objet lui-même**. Python le passe automatiquement : on ne l'écrit pas à l'appel.
- `self.cote = cote` range la valeur du paramètre `cote` dans l'attribut `cote` de l'objet. C'est un **attribut d'instance** : chaque objet a le sien.

````python
class Manuscrit:
    def __init__(self, cote, bibliotheque, date):
        self.cote = cote
        self.bibliotheque = bibliotheque
        self.date = date


manuscrit = Manuscrit("MS 24", "Aberdeen, University Library", 1200)
print(manuscrit.cote)
print(manuscrit.date)
````

Une classe permet de créer autant d'objets que l'on veut. Le `corpus` devient une liste d'objets :

````python
corpus = [
    manuscrit,
    Manuscrit("MS. Ashmole 1511", "Oxford, Bodleian Library", 1210),
]

for ms in corpus:
    print(ms.cote, ":", ms.bibliotheque)
````

Deux erreurs fréquentes : décommentez chaque ligne pour lire le message, puis recommentez-la.

````python
# Un objet ne s'interroge pas avec des crochets, comme un dictionnaire, mais avec un point
# manuscrit["cote"]
# TypeError: 'Manuscrit' object is not subscriptable

# Il manque des arguments : self n'est pas compté, Python le fournit
# Manuscrit("MS 24")
# TypeError: Manuscrit.__init__() missing 2 required positional arguments: 'bibliotheque' and 'date'
````

### **4.2 Les méthodes**

Une **méthode** est une fonction définie dans la classe, avec `self` comme premier paramètre. Elle a ainsi accès aux attributs de l'objet. La fonction `cote_complete(manuscrit)` de la séance 2 devient une méthode :

````python
class Manuscrit:
    def __init__(self, cote, bibliotheque, date):
        self.cote = cote
        self.bibliotheque = bibliotheque
        self.date = date

    def cote_complete(self):
        """Renvoyer la bibliothèque et la cote du manuscrit."""
        return f"{self.bibliotheque}, {self.cote}"


manuscrit = Manuscrit("MS 24", "Aberdeen, University Library", 1200)
corpus = [
    manuscrit,
    Manuscrit("MS. Ashmole 1511", "Oxford, Bodleian Library", 1210),
]

for ms in corpus:
    print(ms.cote_complete())
````

À l'appel, on n'écrit pas `self` : Python passe l'objet lui-même. Les deux lignes ci-dessous sont équivalentes.

````python
print(manuscrit.cote_complete())
print(Manuscrit.cote_complete(manuscrit))   # ce que Python exécute réellement
````

**Attention** : quand on modifie une classe, il faut exécuter à nouveau sa cellule, puis **recréer les objets**. Les objets créés auparavant gardent l'ancienne version de la classe.

### **4.3 Les attributs de classe**

Un **attribut de classe** est défini directement dans la classe, en dehors des méthodes. Il est **partagé par toutes les instances** : on l'utilise pour une valeur commune à tous les objets.

````python
class Manuscrit:
    genre = "bestiaire"

    def __init__(self, cote, bibliotheque, date):
        self.cote = cote
        self.bibliotheque = bibliotheque
        self.date = date


corpus = [
    Manuscrit("MS 24", "Aberdeen, University Library", 1200),
    Manuscrit("MS. Ashmole 1511", "Oxford, Bodleian Library", 1210),
]

# Accès par la classe ou par une instance
print(Manuscrit.genre)
print(corpus[0].genre)
print(corpus[1].genre)
````

**Modifier par la classe** : tous les objets changent.

````python
Manuscrit.genre = "bestiaire latin"
print(corpus[0].genre)
print(corpus[1].genre)
````

**Modifier par une instance** : un seul objet change. L'affectation ne modifie pas l'attribut de classe : elle crée un **attribut d'instance** du même nom, propre à cet objet, qui masque l'attribut de classe.

````python
corpus[0].genre = "bestiaire enluminé"
print(corpus[0].genre)
print(corpus[1].genre)
print(Manuscrit.genre)
````
